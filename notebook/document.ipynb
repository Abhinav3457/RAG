```python
sample_text={
    "../data/text-files/python-intro.txt":"""# Introduction to Python and Machine Learning

## 1. Introduction to Python

Python is a high-level, interpreted, general-purpose programming language known for its simple syntax and readability. It is widely used in software development, web development, automation, data analysis, artificial intelligence, and machine learning.

Python provides several useful features, including:
- Simple and readable syntax.
- Dynamic typing.
- Object-oriented and functional programming support.
- Large collection of built-in libraries.
- Cross-platform compatibility.

Some popular Python libraries include NumPy, Pandas, Matplotlib, Scikit-learn, TensorFlow, and PyTorch.

## 2. Introduction to Machine Learning

Machine Learning (ML) is a branch of Artificial Intelligence (AI) that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed for every task.

Machine learning algorithms analyze historical data, identify relationships, and use those patterns to predict outcomes on new data.

For example, an email spam detection system learns from previously classified emails to identify whether a new email is spam or legitimate.

## 3. Types of Machine Learning

There are three major types of machine learning:

**1. Supervised Learning:** The model learns from labelled datasets where the correct output is already known. Examples include spam detection, house price prediction, and student result prediction.

**2. Unsupervised Learning:** The model works with unlabelled data and identifies hidden patterns or groups. Examples include customer segmentation and clustering.

**3. Reinforcement Learning:** An agent learns by interacting with an environment and receiving rewards or penalties for its actions. It is commonly used in robotics and game-playing systems.

## 4. Python in Machine Learning

Python plays an important role in machine learning because it provides powerful libraries that simplify data processing, model development, training, and evaluation.

Some commonly used libraries are:

- NumPy: Used for numerical calculations and working with arrays.
- Pandas: Used for data manipulation and analysis.
- Matplotlib: Used for creating graphs and visualizations.
- Scikit-learn: Provides machine learning algorithms for classification, regression, and clustering.
- TensorFlow: Used for developing and training deep learning models.
- PyTorch: Used for deep learning and neural network development.

## 5. Machine Learning Workflow

A typical machine learning project follows these steps:

1. Data Collection: Gather relevant data from different sources.
2. Data Preprocessing: Clean the data and handle missing values.
3. Feature Selection: Identify important variables.
4. Model Training: Train the algorithm using available data.
5. Model Evaluation: Measure model performance using suitable metrics.
6. Prediction: Use the trained model to predict results on new data.

## 6. Applications of Machine Learning

Machine learning is used in many real-world applications, including healthcare diagnosis, fraud detection, recommendation systems, facial recognition, autonomous vehicles, financial forecasting, and natural language processing.

## Conclusion

Python and Machine Learning are important technologies in modern software development. Python provides the tools and libraries required to build machine learning applications efficiently. As data continues to grow, machine learning will play an increasingly important role in automation, intelligent systems, and decision-making."""
}
for filepath,content in sample_text.items():
    with open(filepath,'w',encoding='utf-8') as f:
        f.write(content)
print("python-intro file is created successfully and content is written perfectly inside it.")
```
